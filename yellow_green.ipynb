```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
yellow_square = cv2.imread('yellow_square.png',cv2.IMREAD_COLOR)
green_ring = cv2.imread('green_circle.png',cv2.IMREAD_COLOR)
print("Yellow square shape:", yellow_square.shape)
print("Green ring shape:", green_ring.shape)
plt.figure()
plt.imshow(cv2.cvtColor(yellow_square, cv2.COLOR_BGR2RGB))
plt.title("Yellow Square")
plt.show()
plt.figure()
plt.imshow(cv2.cvtColor(green_ring, cv2.COLOR_BGR2RGB))
plt.title("Green Ring")
plt.show()




```

```python
dim = (1000,1000)
yellow_square_dimension = cv2.resize(yellow_square,dim)
print("Resized yellow square shape:", yellow_square_dimension.shape)

```

```python
img_gray_circle = cv2.cvtColor(green_ring, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray_circle, cmap='gray')
plt.title("GRAY SCALE IMAGE FOR GREEN CIRCLE")
```

```python
retval, img_ring_mask = cv2.threshold(img_gray_circle, 127, 255, cv2.THRESH_BINARY)
```

```python
img_ring_mask_inv = cv2.bitwise_not(img_ring_mask)
```

```python
img_combined = cv2.bitwise_and(yellow_square_dimension, yellow_square_dimension, mask=img_ring_mask_inv)
```

```python
plt.figure(figsize=[20, 10])
plt.subplot(141); plt.imshow(img_gray_circle, cmap='gray'); plt.title('Gray Scale Image')
plt.subplot(142); plt.imshow(img_ring_mask, cmap='gray'); plt.title('Mask')
plt.subplot(143); plt.imshow(img_ring_mask_inv, cmap='gray'); plt.title('Inverse Mask')
plt.subplot(144); plt.imshow(img_combined[:, :, ::-1]); plt.title('Combined')
plt.show()
```
